```python
from pathlib import Path
import pandas as pd

# Locate the repository whether this notebook is run from the repo root,
# notebooks/, or the standard Kaggle clone location.
candidates = [
    Path.cwd(),
    Path.cwd().parent,
    Path('/kaggle/working/repo'),
]

repo = next((p for p in candidates if (p / 'results').is_dir()), None)
if repo is None:
    raise FileNotFoundError('Could not locate repository results directory.')

metrics = repo / 'results' / 'metrics.csv'
samples = repo / 'results' / 'sample_translations.csv'

print('=== METRICS ===')
display(pd.read_csv(metrics))

training_dir = repo / 'nllb-en-as'
time_file = training_dir / 'training_time_minutes.txt'
log_file = training_dir / 'training_log.csv'

if time_file.exists():
    print('\n=== TRAINING TIME ===')
    print(time_file.read_text().strip(), 'minutes')

if log_file.exists():
    log = pd.read_csv(log_file)
    if 'eval_loss' in log:
        print('\n=== VALIDATION LOSS ===')
        print(log[log.eval_loss.notna()][['epoch', 'eval_loss']].to_string(index=False))

print('\n=== SAMPLE TRANSLATIONS ===')
display(pd.read_csv(samples))
```
