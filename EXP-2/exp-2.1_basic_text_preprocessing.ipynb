# 1. Convert text to lowercase. (Also display how many uppercase?)


```python
upper_count=0
for i in content:
    if i.isalpha():
        if i.isupper():
            upper_count+=1
print(upper_count)
            
```

```python
content=content.lower()
```

# 2.Remove punctuation.        (Also display how many different types of punctuations?)


```python
import string

punctuation_set = set(string.punctuation)
print(punctuation_set)
```

```python
punch_count=0
punctuation_set.add("\n")
for i in content:
    if i in punctuation_set:
        content = content.replace(i, '')
        punch_count+=1
print(punch_count)
```

```python
content
```

# 3.Remove numbers.            (Also display how many numbers?)


```python
num_count=0
for i in content:
    if i.isnumeric():
        content = content.replace(i, '')
        num_count+=1
```

```python
content
```

# 4. Remove extra whitespace. (How much extra whitespace is present?)


```python
content
```

```python
import re

extra_space_count = len(re.findall(r'\s{2,}', content))

content = re.sub(r'\s+', ' ', content)

print(f"Extra whitespace count: {extra_space_count}")
```
